```python
import rmllm
import psychscanner as psy
import json
import copy
```

```python
rm_sc_st = psy.get_task_template()
rm_2op = rm_sc_st.copy()
rm_2op_tc = rm_sc_st.copy()


rm_2op["taskname"] = "zs_rm_2op"
rm_2op_tc["taskname"] = "zs_rm_2op_tc"

rm_2op["contexts"] = ["perceived","imagined"]
rm_2op_tc["contexts"] = ["perceived","imagined"]


rm_2op["contexts_id"] = ["perceived","imagined"]
rm_2op_tc["contexts_id"] = ["perceived","imagined"]


rm_2op["context_present"] = False
rm_2op_tc["context_present"] = False


rm_2op["chain_type"] = "item"
rm_2op_tc["chain_type"] = "trial"
```

```python
task_prep_dir = rmllm.config.RAW_DATA_DIR/"prepare_task"
with open(task_prep_dir/"task_rm_v1.json","r") as f:
    taskbase_2op = json.load(f)

with open(task_prep_dir/"rm_task_2op_inst.json") as f:
    instr_2op = json.load(f)
```

```python
rm_2op["instructions"] = instr_2op["order1"]
rm_2op["items"] = {}
# Fill rm_2op
for i, j in enumerate(taskbase_2op["task1"]["trials"]):
    ttype = j["source"].split(" ")[-1].lower()
    tr_code = f"{ttype}_{i+1}"
    rm_2op["items"][tr_code] = []
    rm_2op["items"][tr_code].append(
        {
            "stimulus": {
                "Word_Pair": {'word_1': j['stim']['word 1'],
                'word_2': j['stim']['word 2']},
                },
                "trcode": tr_code
            }
        )


rm_2op_r = copy.deepcopy(rm_2op)
```

```python
# Reverse options for rm2op
rm_2op_r["instructions"]["INSTRUCTIONS"]["JUDGE_GENERATION_TYPE_OF_WORD_2"] = instr_2op["order2"]["JUDGE_GENERATION_TYPE_OF_WORD_2"]
```

```python
rm_2op_tc["instructions"] = {}
rm_2op_tc["instructions"] = {"definition":rm_2op["instructions"]["definition"]}

rm_2op_tc["items"] = {}
# Fill rm_2op_tc items
for i, j in enumerate(taskbase_2op["task1"]["trials"]):

    ttype = j["source"].split(" ")[-1].lower()
    tr_code = f"{ttype}_{i+1}"
    rm_2op_tc["items"][tr_code] = []
    rm_2op_tc["items"][tr_code].append(
        {
            "stimulus": {
                "instructions": {"WORD_PAIR_TASK":
                                 instr_2op['order1']["INSTRUCTIONS"]['WORD_PAIR_TASK']},
                "Word_Pair": {'word_1': j['stim']['word 1'], 'word_2': j['stim']['word 2']},
                "Query": "Please report the **Word_2** as per instructions. Word 2 = "
            },
            "trcode": tr_code
        }
    )
    rm_2op_tc["items"][tr_code].append(
        {
            "stimulus": {
                "instructions": 
                {"RELATEDNESS_RATING_BETWEEN_WORD_1_AND_WORD_2":instr_2op['order1']["INSTRUCTIONS"]["RELATEDNESS_RATING_BETWEEN_WORD_1_AND_WORD_2"]},
                "Query": "Please report **Relatedness Rating** between the two words in the earlier word-pair as per instructions. Relatedness Rating = "
            },
            "trcode": tr_code
        }
    )
    rm_2op_tc["items"][tr_code].append(
            {
                "stimulus": {
                    "instructions": 
                    {"JUDGE_GENERATION_TYPE_OF_WORD_2": instr_2op['order1']["INSTRUCTIONS"]["JUDGE_GENERATION_TYPE_OF_WORD_2"]},
                    "Query": "Please select and report the **Judgment** of the word 2 value regarding genration type using the earlier word-pair. Select one of the generation type options as per the instructions. Judgment = "
                },
                "trcode": tr_code
            }
        )
    rm_2op_tc["items"][tr_code].append(
            {
                "stimulus": {
                    "instructions": 
                    {"CONFIDENCE_ON_JUDGEMENT_OF_GENERATION_TYPE":instr_2op['order1']["INSTRUCTIONS"]["CONFIDENCE_ON_JUDGEMENT_OF_GENERATION_TYPE"]},
                    "Query": "Please report the **Confidence** on your judgment being **correct** as per the confidence rating scale insturction in your generation type judgment ealier. Confidence = "
                },
                "trcode": tr_code
            }
        )
```

```python
rm_2op_tc_r["instructions"] = {}
rm_2op_tc_r["instructions"] = {"definition":rm_2op_r["instructions"]["definition"]}

rm_2op_tc_r["items"] = {}
# Fill rm_2op_tc items
for i, j in enumerate(taskbase_2op["task1"]["trials"]):

    ttype = j["source"].split(" ")[-1].lower()
    tr_code = f"{ttype}_{i+1}"
    rm_2op_tc_r["items"][tr_code] = []
    rm_2op_tc_r["items"][tr_code].append(
        {
            "stimulus": {
                "instructions": {"WORD_PAIR_TASK":
                                 instr_2op['order1']["INSTRUCTIONS"]['WORD_PAIR_TASK']},
                "Word_Pair": {'word_1': j['stim']['word 1'], 'word_2': j['stim']['word 2']},
                "Query": "Please report the **Word_2** as per instructions. Word 2 = "
            },
            "trcode": tr_code
        }
    )
    rm_2op_tc_r["items"][tr_code].append(
        {
            "stimulus": {
                "instructions": 
                {"RELATEDNESS_RATING_BETWEEN_WORD_1_AND_WORD_2":instr_2op['order1']["INSTRUCTIONS"]["RELATEDNESS_RATING_BETWEEN_WORD_1_AND_WORD_2"]},
                "Query": "Please report **Relatedness Rating** between the two words in the earlier word-pair as per instructions. Relatedness Rating = "
            },
            "trcode": tr_code
        }
    )
    rm_2op_tc_r["items"][tr_code].append(
            {
                "stimulus": {
                    "instructions": 
                    {"JUDGE_GENERATION_TYPE_OF_WORD_2": instr_2op['order2']["JUDGE_GENERATION_TYPE_OF_WORD_2"]},
                    "Query": "Please select and report the **Judgment** of the word 2 value regarding genration type using the earlier word-pair. Select one of the generation type options as per the instructions. Judgment = "
                },
                "trcode": tr_code
            }
        )
    rm_2op_tc_r["items"][tr_code].append(
            {
                "stimulus": {
                    "instructions": 
                    {"CONFIDENCE_ON_JUDGEMENT_OF_GENERATION_TYPE":instr_2op['order1']["INSTRUCTIONS"]["CONFIDENCE_ON_JUDGEMENT_OF_GENERATION_TYPE"]},
                    "Query": "Please report the **Confidence** on your judgment being **correct** as per the confidence rating scale insturction in your generation type judgment ealier. Confidence = "
                },
                "trcode": tr_code
            }
        )
```

```python
rm_2op["parser"] = "ResponseRmStEI"
rm_2op_r["parser"] = "ResponseRmStIE"


rm_2op_tc["parser"] = "AllResponseRMEI"
rm_2op_tc_r["parser"] = "AllResponseRMIE"

```

```python
# save files
def savejson(dir,name,data):
    with open(dir/name,"w") as f:
        json.dump(data,f,indent=4)

OUT_DIR = rmllm.config.RAW_DATA_DIR/"runtasks2"
if not OUT_DIR.is_dir():
    OUT_DIR.mkdir()

savejson(OUT_DIR,"rm_2op.json",rm_2op)
savejson(OUT_DIR,"rm_2op_r.json",rm_2op_r)
savejson(OUT_DIR,"rm_2op_tc.json",rm_2op_tc)
savejson(OUT_DIR,"rm_2op_tc_r.json",rm_2op_tc_r)
```
