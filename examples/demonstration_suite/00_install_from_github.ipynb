# Installing psychscanner from GitHub

Mirrors the [README Installation section](../README.md#installation): create a `uv` environment, clone the repo, install in editable mode, then verify the import.

## 0. Install uv

Skip this if `uv` is already on your PATH (check with `uv --version`).

```python
%%bash
# macOS / Linux
curl -LsSf https://astral.sh/uv/install.sh | sh

# Windows (PowerShell), instead:
# powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"

# Already using conda? Install uv into your base env instead:
# conda install -c conda-forge uv
```

## 1. Create the uv environment

```python
%%bash
uv venv psyscan --python 3.11
source psyscan/bin/activate
```

## 2. Clone and install psychscanner

```python
%%bash
git clone https://github.com/saurabhr/psychscanner.git
cd psychscanner
uv pip install -e .
```

Alternatively, install directly from GitHub without a local clone:

```python
%%bash
uv pip install git+https://github.com/saurabhr/psychscanner.git
```

## 3. Verify the install

```python
import psychscanner

print(psychscanner.__file__)
```
